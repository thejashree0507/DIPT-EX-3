### Write a Python program using OpenCV to perform histogram equalization on the given grayscale image **"parrot.jpg"**.

The program should:

- Import the required libraries (**OpenCV, NumPy, Matplotlib**).
- Read the image **"parrot.jpg"** in **grayscale** format.
- Display the grayscale image and plot its histogram.
- Perform **histogram equalization** on the grayscale image using `cv2.equalizeHist()` to enhance its contrast.
- Using Matplotlib, display the following in a **2 × 2 grid layout**:
  - Original Grayscale Image
  - Histogram of Original Image
  - Enhanced (Equalized) Image
  - Histogram of Enhanced Image


```python
# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image in grayscale format
img=cv2.imread("parrot.jpg",0)
```

```python
# Display the grayscale image.
plt.imshow(img,cmap='gray')
plt.title("parrot image")
plt.axis("off")
plt.show()
```

```python
# Plot the histogram of the grayscale image
plt.hist(img.ravel(),256,range=[0,255])
plt.title("histogram of grayscale image")
plt.show()
```

```python
# Perform histogram equalization
img_eq=cv2.equalizeHist(img)
```

```python
plt.imshow(img_eq, cmap='gray')
plt.title('Enhanced Image')
plt.show()
```

```python
plt.hist(img_eq.ravel(), 256, range = [0, 256])
plt.title('Equalized Histogram')
```

```python
# Display [1] the Original Image (Gray Image) and its Histogram, and [2] the Enhanced Image and its Histogram using a 2×2 layout in Matplotlib.
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1);plt.imshow(img, cmap='gray');plt.title("Original Image");plt.axis("off")
plt.subplot(2, 2, 2);plt.hist(img.ravel(), bins=256, range=[0, 256]);plt.title("Original Histogram");plt.xlim([0, 256])
plt.subplot(2, 2, 3);plt.imshow(img_eq, cmap='gray');plt.title("Equalized Image");plt.axis("off")
plt.subplot(2, 2, 4);plt.hist(img_eq.ravel(), bins=256, range=[0, 256]);plt.title("Equalized Histogram");plt.xlim([0, 256])
plt.tight_layout()
plt.show()
```
